```python

# Put data into a DataFrame
df_mg = pd.DataFrame({"middate": pd.to_datetime(middate),"MgCa": mg2})


# Group by date: one mean & std per middate
mg_stats = df_mg.groupby("middate")["MgCa"].agg(["mean", "std"]).reset_index()

x_mg    = mg_stats["middate"]       # same length
y_mg    = mg_stats["mean"]          # same length
yerr_mg = mg_stats["std"]           # same length

# Put data into a DataFrame
df_u = pd.DataFrame({"middate": pd.to_datetime(middate),"UCa": u2}).dropna(subset=["middate", "UCa"])

# Group by date: one mean & std per middate
u_stats = df_u.groupby("middate")["UCa"].agg(["mean", "std"]).reset_index()

x_u    = u_stats["middate"]
y_u    = u_stats["mean"]
yerr_u = u_stats["std"]


fig, ax = plt.subplots(figsize=(10,6))

twin1 = ax.twinx()
twin2 = ax.twinx()
twin2.spines.right.set_position(("axes", 1.2))

# Mg/Ca suite points with std
mg_err = twin1.errorbar(x_mg, y_mg, yerr=yerr_mg,fmt='o', capsize=4, label="Mg/Ca (±SD)", color = 'g')

# U/Ca suite points with std
u_err = twin2.errorbar(x_u, y_u, yerr=yerr_u,fmt='o', capsize=4, label="U/Ca (±SD)", color ='b')

# Discharge (ft^3/s) (3-day avg)
day3.scatter = ax.scatter(day3, f3, label="3-Day Discharge (ft^3/s) Avg", color = 'r')
day3.line, = ax.plot(day3, f3, label="3-Day Discharge (ft^3/s) Avg", color = 'r')

# Labels
ax.set_xlabel("Time (Year-Month)")
ax.set_ylabel("Discharge (ft^3/s)")
twin1.set_ylabel("Mg/Ca (mmol/mol)")
twin2.set_ylabel("U/Ca (nmol/mol)")

# Legend
handles = [day3.scatter, mg_err, u_err]
labels = [h.get_label() for h in handles]
ax.legend(handles, labels, loc="upper left")

plt.title('3-Day Average Discharge (ft^3/s) of Columbia River vs Mg/Ca and U/Ca Averages')

plt.tight_layout()
plt.show()
```
