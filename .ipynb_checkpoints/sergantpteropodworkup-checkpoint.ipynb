```python
#Convert object columns to numeric data
Mg_CaSER = pd.to_numeric(Mg_CaSER, errors='coerce')
Mn_CaSER = pd.to_numeric(Mn_CaSER, errors='coerce')
Fe_CaSER = pd.to_numeric(Fe_CaSER, errors='coerce')
Al_CaSER = pd.to_numeric(Al_CaSER, errors='coerce')
Sr_CaSER = pd.to_numeric(Sr_CaSER, errors='coerce')
```

```python
#graph 5: create subplots w linear regression lines for Mg/Ca vs Mn/Ca, Fe/Ca vs Al/ca

#create figure
fig, (axes) = plt.subplots(nrows=1, ncols=2, figsize=(15,6))

#linear regression lines
#Mg/Ca vs Mn/Ca
#sergent data
slopeMnMgSER, interceptMnMgSER, r_valueMnMgSER, p_valueMnMgSER, std_errMnMgSER = linregress(Mg_CaSER, Mn_CaSER)
regression_yMnMgSER = slopeMnMgSER * Mg_CaSER + interceptMnMgSER
r_squaredMnMgSER = r_valueMnMgSER**2
#red+sonic sediment pteropods
slopeMnMgRSED, interceptMnMgRSED, r_valueMnMgRSED, p_valueMnMgRSED, std_errMnMgRSED = linregress(redsonicMg, redsonicMn)
regression_yMnMgRSED = slopeMnMgRSED * redsonicMg + interceptMnMgRSED
r_squaredMnMgRSED = r_valueMnMgRSED**2
#ox+red+sonic sediment pteropods
slopeMnMgOSED, interceptMnMgOSED, r_valueMnMgOSED, p_valueMnMgOSED, std_errMnMgOSED = linregress(oxredsonicMg, oxredsonicMn)
regression_yMnMgOSED = slopeMnMgOSED * oxredsonicMg + interceptMnMgOSED
r_squaredMnMgOSED = r_valueMnMgOSED**2
#sediment trap pteropods
slopeMnMgW, interceptMnMgW, r_valueMnMgW, p_valueMnMgW, std_errMnMgW = linregress(Mg_Ca, Mn_Ca)
regression_yMnMgW = slopeMnMgW * Mg_Ca + interceptMnMgW
r_squaredMnMgW = r_valueMnMgW**2

#Al/Ca vs Fe/Ca
#sergent data
slopeAlFeSER, interceptAlFeSER, r_valueAlFeSER, p_valueAlFeSER, std_errAlFeSER = linregress(Fe_CaSER, Al_CaSER)
regression_yAlFeSER = slopeAlFeSER * Fe_CaSER + interceptAlFeSER
r_squaredAlFeSER = r_valueAlFeSER**2
#red+sonic sediment pteropods
slopeAlFeRSED, interceptAlFeRSED, r_valueAlFeRSED, p_valueAlFeRSED, std_errAlFeRSED = linregress(redsonicFe, redsonicAl)
regression_yAlFeRSED = slopeAlFeRSED * redsonicFe + interceptAlFeRSED
r_squaredAlFeRSED = r_valueAlFeRSED**2
#ox+red+sonic pteropds
slopeAlFeOSED, interceptAlFeOSED, r_valueAlFeOSED, p_valueAlFeOSED, std_errAlFeOSED = linregress(oxredsonicFe, oxredsonicAl)
regression_yAlFeOSED = slopeAlFeOSED * oxredsonicFe + interceptAlFeOSED
r_squaredAlFeOSED = r_valueAlFeOSED**2
# sediment trap pteropods
slopeAlFeW, interceptAlFeW, r_valueAlFeW, p_valueAlFeW, std_errAlFeW = linregress(Fe_Ca, Al_Ca)
regression_yAlFeW = slopeAlFeW * Fe_Ca + interceptAlFeW
r_squaredAlFeW = r_valueAlFeW**2

#scatter plot data points
axes[0].scatter(Mg_Ca, Mn_Ca, color = 'green')
axes[0].scatter(redsonicMg, redsonicMn, color = 'blue')
axes[0].scatter(oxredsonicMg, oxredsonicMn, color ='orange')
axes[0].scatter(Mg_CaSER, Mn_CaSER, color = 'r')
axes[0].set_xlabel('Mg/Ca mmols/mol')
axes[0].set_ylabel('Mn/Ca umols/Ca')

axes[1].scatter(Fe_Ca, Al_Ca, color = 'green')
axes[1].scatter(redsonicFe, redsonicAl, color = 'blue')
axes[1].scatter(oxredsonicFe, oxredsonicAl, color ='orange')
axes[1].scatter(Fe_CaSER, Al_CaSER, color = 'r')
axes[1].set_xlabel('Fe/Ca umols/mol')
axes[1].set_ylabel('Al/Ca umols/mol')

#plot linear regression

#Mg/Ca vs Mn/Ca
#sergent data
axes[0].plot(Mg_CaSER, regression_yMnMgSER, label=f'Sergent Mg/Ca vs Mn/Ca (R²={r_squaredMnMgSER:.2f})), (p={p_valueMnMgSER:.2e}), (slope={slopeMnMgSER:.2f})', color='r', linestyle='--')
#red+sonic sedimentary 
axes[0].plot(redsonicMg, regression_yMnMgRSED, label=f'Red+Sonic Sedimentary Mg/Ca vs Mn/Ca (R²={r_squaredMnMgRSED:.2f})), (p={p_valueMnMgRSED:.2e}), (slope={slopeMnMgRSED:.2f})', color='b', linestyle='--')
#ox+red+sonic sedimentary 
axes[0].plot(oxredsonicMg, regression_yMnMgOSED, label=f'Ox+Red+Sonic Sedimentary Mg/Ca vs Mn/Ca (R²={r_squaredMnMgOSED:.2f})), (p={p_valueMnMgOSED:.2e}), (slope={slopeMnMgOSED:.2f})', color='orange', linestyle='--')
#sediment pteropods 
axes[0].plot(Mg_Ca, regression_yMnMgW, label=f'Sediment Trap Mg/Ca vs Mn/Ca (R²={r_squaredMnMgW:.2f})), (p={p_valueMnMgW:.2e}), (slope={slopeMnMgW:.2f})', color='green', linestyle='--')

#Fe/Ca vs Al/Ca
#sergent data
axes[1].plot(Fe_CaSER, regression_yAlFeSER, label=f'Sergent Fe/Ca vs Al/Ca (R²={r_squaredAlFeSER:.2f})), (p={p_valueAlFeSER:.2e}), (slope={slopeAlFeSER:.2f})', color='r', linestyle='--')
#red+sonic sedimentary 
axes[1].plot(redsonicFe, regression_yAlFeRSED, label=f'Red+Sonic Sedimentary Fe/Ca vs Al/Ca (R²={r_squaredAlFeRSED:.2f})), (p={p_valueAlFeRSED:.2e}), (slope={slopeAlFeRSED:.2f})', color='b', linestyle='--')
#ox+red+sonic sedimentary 
axes[1].plot(oxredsonicFe, regression_yAlFeOSED, label=f'Ox+Red+Sonic Sedimentary Fe/Ca vs Al/Ca (R²={r_squaredAlFeOSED:.2f})), (p={p_valueAlFeOSED:.2e}), (slope={slopeAlFeOSED:.2f})', color='orange', linestyle='--')
#sediment pteropods 
axes[1].plot(Fe_Ca, regression_yAlFeW, label=f'Sediment Trap Fe/Ca vs Al/Ca (R²={r_squaredAlFeW:.2f})), (p={p_valueAlFeW:.2e}), (slope={slopeAlFeW:.2f})', color='green', linestyle='--')


fig.suptitle('Detrial and Continential Influence on Sergent and Pavia Pteropods', fontsize = 20)

label = 'Sediment Trap', 'Red+Sonic Sedimentary', 'Ox+Red+Sonic Sedimentary', 'Sergent Pterpods'
#fig.legend(label, bbox_to_anchor=(1.02, 0.5), loc='center left', borderaxespad=0)
fig.legend(bbox_to_anchor=(1.02, 0.5), loc='center left', title='Pteropod Type', borderaxespad=0)
# axes[1].legend(bbox_to_anchor = (0.5,1))
# axes[0].legend(bbox_to_anchor = (0.5,1))
# Adjusting layout for better spacing
fig.tight_layout()

```

```python
#U/Ca vs Mg/Ca with linear regressions

#create figure
plt.figure(figsize=(10, 6))

#scatter data
plt.scatter(Mg_Ca,Sr_Ca, color = 'green')
plt.scatter(redsonicMg, redsonicSr,  color = 'blue')
plt.scatter(oxredsonicMg, oxredsonicSr, color = 'orange')
plt.scatter(Mg_CaSER, Sr_CaSER, color = 'r')

#calculate linear regressions 
#sergent data
slopeSrMgSER, interceptSrMgSER, r_valueSrMgSER, p_valueSrMgSER, std_errSrMgSER = linregress(Mg_CaSER, Sr_CaSER)
regression_ySrMgSER = slopeSrMgSER * Mg_CaSER + interceptSrMgSER
r_squaredSrMgSER = r_valueSrMgSER**2

#red+sonic sediment pteropods
slopeSrMgRSED, interceptSrMgRSED, r_valueSrMgRSED, p_valueSrMgRSED, std_errSrMgRSED = linregress(redsonicMg, redsonicSr)
regression_ySrMgRSED = slopeSrMgRSED * redsonicMg + interceptSrMgRSED
r_squaredSrMgRSED = r_valueSrMgRSED**2

#ox+red+sonic sediment pteropods
slopeSrMgOSED, interceptSrMgOSED, r_valueSrMgOSED, p_valueSrMgOSED, std_errSrMgOSED = linregress(oxredsonicMg, oxredsonicSr)
regression_ySrMgOSED = slopeSrMgOSED * oxredsonicMg + interceptSrMgOSED
r_squaredSrMgOSED = r_valueSrMgOSED**2

#sediment trap pteropods
slopeSrMgW, interceptSrMgW, r_valueSrMgW, p_valueSrMgW, std_errSrMgW = linregress(Mg_Ca, Sr_Ca)
regression_ySrMgW = slopeSrMgW * Mg_Ca + interceptSrMgW
r_squaredSrMgW = r_valueSrMgW**2

#plot linear regression
#Mg/Ca vs Sr/Ca
#sergent data
plt.plot(Mg_CaSER, regression_ySrMgSER, label=f'Sergent (R²={r_squaredSrMgSER:.2f})), (p={p_valueSrMgSER:.2e}), (slope={slopeSrMgSER:.2f})', color='r', linestyle='--')
#red+sonic sedimentary 
plt.plot(redsonicMg, regression_ySrMgRSED, label=f'Red+Sonic Sedimentary (R²={r_squaredSrMgRSED:.2f})), (p={p_valueSrMgRSED:.2e}), (slope={slopeSrMgRSED:.2f})', color='b', linestyle='--')
#ox+red+sonic sedimentary 
plt.plot(oxredsonicMg, regression_ySrMgOSED, label=f'Ox+Red+Sonic Sedimentary (R²={r_squaredSrMgOSED:.2f})), (p={p_valueSrMgOSED:.2e}), (slope={slopeSrMgOSED:.2f})', color='orange', linestyle='--')
#sediment pteropods 
plt.plot(Mg_Ca, regression_ySrMgW, label=f'Sediment Trap (R²={r_squaredSrMgW:.2f})), (p={p_valueSrMgW:.2e}), (slope={slopeSrMgW:.2f})', color='green', linestyle='--')


plt.xlabel('Mg/Ca mmols/mol')
plt.ylabel('Sr/Ca mmols/mol')
plt.title ('Sr/Ca vs Mg/Ca for All Pteropod Types')
label = 'Sediment Trap', 'Red+Sonic Sedimentary', 'Ox+Red+Sonic Sedimentary', 'Sergent Pteropods'
plt.legend(bbox_to_anchor=(1.02, 0.5))
```
